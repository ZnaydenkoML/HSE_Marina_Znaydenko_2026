```python
import datetime as dt

class Contract:
    def __init__(self, contract_number, creditor, debtor, amount, deadline):
        self.contract_number = contract_number
        self.creditor = creditor
        self.debtor = debtor
        self.amount = amount
        self.deadline = deadline
        self.is_signed = False
        self.is_fulfilled = False
        self.is_terminated = False
        self.termination_reason = ""
        self.amendments = []

    def sign_contract(self):
        if self.is_signed:
            raise ValueError('Договор уже подписан')
        self.is_signed = True

    def add_amendment(self, description, date):
        if not self.is_signed:
            raise ValueError('Нельзя изменить неподписанный договор')
        if self.is_terminated:
            raise ValueError('Договор расторгнут, изменения невозможны')
        self.amendments.append({'description': description, 'date': date})

    def fulfill_obligation(self):
        if not self.is_signed:
            raise ValueError('Договор не подписан')
        if self.is_terminated:
            raise ValueError('Договор расторгнут')
        self.is_fulfilled = True

    def terminate_contract(self, reason=""):
        self.is_terminated = True
        self.termination_reason = reason

    def is_overdue(self):
        return not self.is_fulfilled and dt.datetime.now() > self.deadline
```

```python
contract = Contract(
    contract_number='D-2026-001',
    creditor='ООО Ромашка',
    debtor='ИП Иванов',
    amount=500000,
    deadline=dt.datetime(2026, 10, 1)
)

contract.sign_contract()
contract.add_amendment('Продление срока оплаты', dt.datetime(2026, 9, 20))
print(contract.is_overdue())        # False, если сегодняшняя дата раньше deadline
contract.fulfill_obligation()
print(contract.is_fulfilled)        # True
```
