# Задание
Сгенерируйте массив из целых чисел, содержащий 100 000 элементов, с помощью функции randomint из модуля random.

 

Сгенерируйте с помощью функции range массив, содержащий словари со следующей структурой:

 

{

“num_1”: randomint(1,1_000_000),

“num_2”: randomint(1,1_000_000)

}

 

Длина массива должна составлять 100 000 элементов.

 

Напишите функцию алгоритма сортировки пузырьком и с её помощью отсортируйте первый массив.

 

Отсортируйте второй массив с помощью встроенного спиского метода .sort() и лямбда-функции сначала по первому ключу, потом по второму



```python
from random import randint
from time import perf_counter


# Количество элементов
ARRAY_SIZE = 100_000


def bubble_sort(array):
    """
    Сортирует переданный список по возрастанию
    методом пузырьковой сортировки.
    """
    length = len(array)

    for i in range(length - 1):
        swapped = False

        # После каждой итерации самый большой элемент
        # перемещается в конец списка
        for j in range(length - i - 1):
            if array[j] > array[j + 1]:
                array[j], array[j + 1] = array[j + 1], array[j]
                swapped = True

        # Если перестановок не было, список уже отсортирован
        if not swapped:
            break

    return array


def check_numbers_sorting(array):
    """Проверяет сортировку списка чисел."""
    return all(
        array[i] <= array[i + 1]
        for i in range(len(array) - 1)
    )


def check_dictionaries_sorting(array):
    """Проверяет сортировку списка словарей по двум ключам."""
    return all(
        (
            array[i]["num_1"],
            array[i]["num_2"]
        )
        <=
        (
            array[i + 1]["num_1"],
            array[i + 1]["num_2"]
        )
        for i in range(len(array) - 1)
    )


# --------------------------------------------------
# 1. Создание массива из 100 000 случайных чисел
# --------------------------------------------------

numbers = [
    randint(1, 1_000_000)
    for _ in range(ARRAY_SIZE)
]

print("Первый массив создан.")
print("Количество элементов:", len(numbers))
print("Первые 10 элементов до сортировки:")
print(numbers[:10])


# --------------------------------------------------
# 2. Создание массива из 100 000 словарей
# --------------------------------------------------

dictionaries = [
    {
        "num_1": randint(1, 1_000_000),
        "num_2": randint(1, 1_000_000)
    }
    for _ in range(ARRAY_SIZE)
]

print("\nВторой массив создан.")
print("Количество словарей:", len(dictionaries))
print("Первые 5 словарей до сортировки:")

for item in dictionaries[:5]:
    print(item)


# --------------------------------------------------
# 3. Сортировка первого массива методом пузырька
# --------------------------------------------------

print("\nНачалась пузырьковая сортировка первого массива...")

start_time = perf_counter()

bubble_sort(numbers)

bubble_sort_time = perf_counter() - start_time

print("Пузырьковая сортировка завершена.")
print("Время выполнения:", round(bubble_sort_time, 3), "сек.")
print("Первые 10 элементов после сортировки:")
print(numbers[:10])
print(
    "Массив отсортирован правильно:",
    check_numbers_sorting(numbers)
)


# --------------------------------------------------
# 4. Сортировка второго массива методом .sort()
# --------------------------------------------------

print("\nНачалась сортировка массива словарей...")

start_time = perf_counter()

dictionaries.sort(
    key=lambda item: (
        item["num_1"],
        item["num_2"]
    )
)

dictionary_sort_time = perf_counter() - start_time

print("Сортировка массива словарей завершена.")
print("Время выполнения:", round(dictionary_sort_time, 3), "сек.")
print("Первые 10 словарей после сортировки:")

for item in dictionaries[:10]:
    print(item)

print(
    "Словари отсортированы правильно:",
    check_dictionaries_sorting(dictionaries)
)
```
